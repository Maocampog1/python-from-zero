El sistema metro de la ciudad tiene una máquina para reciclar botellas de plástico. Esta máquina otorga puntos reembolsables en servicios del sistema. Por cada botella reciclada al usuario le dan 5 puntos. Además, si el usuario recicla 10 o más botellas, el metro le otorga 20 puntos de bonificación.

Cree un programa en Python donde:
- Pida al usuario por pantalla la cantidad de botellas a reciclar (int).
- Imprima la cantidad de puntos que el usuario se ha ganado.

Ejemplo 1:

Si el usuario ingresa 5, salida por pantalla debe mostrar:

25


Ejemplo 2:

Si el usuario ingresa 10, salida por pantalla debe mostrar:

70

```python
CantidadBotellasReciclar=int(input())
puntos =CantidadBotellasReciclar*5
if CantidadBotellasReciclar>=10:
  puntos+=20
print(puntos)
```
