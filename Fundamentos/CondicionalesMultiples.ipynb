Un grupo de analistas políticos necesita determinar si un partido político ha superado el umbral mínimo necesario para obtener una representación en el Congreso.​

​

 Cree un programa que:​

Pida por pantalla la cantidad de votos válidos en la elección (int).​

Pida por pantalla la cantidad de votos obtenidos por el partido en cuestión (int).​

Valide que la cantidad de votos del partido es mayor al 3% de los votos válidos.​

Si se cumple lo anterior, el programa imprimirá “Tu partido tendrá curules”.​

De lo contrario, el programa imprimirá “Se quemaron”.​

```python
cantidadVotosValidos = int(input("Ingrese la cantidad de votos validos: "))
cantidadVotosPartido = int(input("Ingrese la cantidad de votos del partido: "))
if cantidadVotosPartido > (cantidadVotosValidos*0.03):
  print("Tu partido tendrá curules")
else:
  print("Se quemaron")
```

#El Elif


Pídale al usuario por pantalla la temperatura. Si la temperatura es mayor a 27 imprima el mensaje “Comprar helado”. De lo contrario, si la temperatura es menor a 15 grados imprima “Comprar chocolate”. De lo contrario, imprima “Comprar jugo de naranja”.​

Al final siempre imprima “Fin programa”.​

```python
temperatura = int(input("Ingrese la temperatura: "))
if temperatura >27:
  print("Comprar helado")
elif temperatura <15:
  print("Comprar chocolate")
else:
  print("Comprar jugo de naranja")
print("Fin programa")
```

Imagina que estás en una reunión familiar y alguien menciona que las nuevas generaciones no saben vivir sin internet. Entre risas, decides hacer un programa que determine a qué generación pertenece cada persona según su año de nacimiento.​

El programa pedirá por pantalla el año de nacimiento (int).​

Si el año está entre 1994 y 2010, imprimirá “Eres Generación Z”.​

Si no, si el año está entre 1981 y 1993 imprimirá “Eres Milennial”.​

De lo contrario, imprimirá “Eres de otra generación”.​

```python
añoDeNacimiento = int(input("Ingrese su año de nacimiento: "))
if añoDeNacimiento >= 1994 and añoDeNacimiento <= 2010
  print("Eres Generación Z")
elif añoDeNacimiento >= 1981 and añoDeNacimiento <= 1993:
  print("Eres Milennial")
else:
  print("Eres de otra Generacion")
```
