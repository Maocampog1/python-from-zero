En una oficina de cambio de dinero sólo permiten cambiar entre 200 y 2000 dólares a pesos colombianos.

Cree un programa en Python para calcular la cantidad de pesos colombianos equivalente a los dólares que se quieren cambiar o que indique si no se puede realizar el cambio.

En el programa:
- Pida al usuario la cantidad de dólares a cambiar (float).
- Si la cantidad de dólares está entre 200 y 2000 realicé el cambio a pesos colombianos con el valor del dólar en 3900. Luego imprima "COP" seguido por un espacio y luego el valor en pesos colombianos cambiado. En caso contrario, imprima No se puede realizar el cambio.

 Si el usuario ingresa 350, luego el programa debe mostrar:

COP 1365000

 Si el usuario ingresa 2100, el programa debe mostrar:

Dos candidatos están participando en las elecciones presidenciales. Después de las votaciones, necesitamos conocer cuál fue el candidato ganador o si hubo empate. Cree un programa en Python que: le pida al usuario el nombre del primer candidato (nombre_del_candidato1) (str), después pida la cantidad de votos del primer candidato (int), luego pida el nombre del segundo candidato (nombre_del_candidato2 ) (str), y después pida la cantidad de votos del segundo candidato (int).

Finalmente imprima cuál candidato es ganador o si hay empate.
- Si el ganador es el candidato 1 imprima El presidente electo es nombre_del_candidato1
- Si es el candidato 2 imprima El presidente electo es nombre_del_candidato2
- Si ambos candidatos tienen la misma cantidad de votos imprima Hay empate

Ejemplo:

Si el usuario ingresa Jhon Perez 20 y luego Luis Portocarreño 30, la salida por pantalla es:

El presidente electo es Luis Portocarreño

```python
nombreCandidaro1=str(input("Ingrese el nombre del primer candidato: "))
cantidadVotosCandidato1=int(input("Ingrese la cantidad de votos del primer candidato: "))
nombreCandidaro2=str(input("Ingrese el nombre del segundo candidato: "))
cantidadVotosCandidato2=int(input("Ingrese la cantidad de votos del segundo candidato: "))
if cantidadVotosCandidato1 > cantidadVotosCandidato2:
  print(f"El presidente electo es {nombreCandidaro1}")
elif cantidadVotosCandidato1 < cantidadVotosCandidato2:
  print(f"El presidente electo es {nombreCandidaro2}")
elif cantidadVotosCandidato1 == cantidadVotosCandidato2:
  print("Hay empate")
```

El mercado de divisas permite a los inversionistas realizar las siguientes acciones:

Comprar divisas
Vender divisas
Realice un programa en Python que represente el funcionamiento del mercado de divisas, en el cual:

Pida al usuario la cantidad de divisas que posee (int).
Pida al usuario la acción a realizar (1 para comprar divisas o 2 para vender divisas) (int).
Si la acción ingresada es 1, pida la cantidad de divisas a comprar (int) y sume ese valor ingresado a las divisas que posee el usuario.
Si la acción ingresada es 2, pida la cantidad de divisas a vender (int), y reste ese valor ingresado a las divisas que posee el usuario.
Siempre, al final, imprima la cantidad de divisas con las que quedó el usuario.
Ejemplo 1:

Si el usuario ingresa 50, luego ingresa 2, luego ingresa 30, el programa debe mostrar:

20

Ejemplo 2:

Si el usuario ingresa  50, luego ingresa 1, luego ingresa 50, el programa debe mostrar:

100

```python
cantidadDivisas = int(input("Ingrese la cantidad de divisas que posee: "))
accion=int(input("Ingrese la acción a realizar (1 para comprar divisas o 2 para vender divisas): "))
if accion == 1:
  cantidadDivisasAComprar = int(input("Ingrese la cantidad de divisas a comprar: "))
  cantidadDivisas = cantidadDivisas + cantidadDivisasAComprar
  print(cantidadDivisas)
elif  accion == 2:
  cantidadDivisasAVender = int(input("Ingrese la cantidad de divisas a vender: "))
  cantidadDivisas = cantidadDivisas - cantidadDivisasAVender
  print(cantidadDivisas)
```
