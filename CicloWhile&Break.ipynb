#**Ciclos** **While**

Simule un semáforo en rojo que se demora 5 segundos para cambiar. Use una variable de control llamada segundos que empieza en 0 e incrementa de 1 en 1 cada vez que se repite el ciclo. Mientras segundos sea menor que 5, se muestra en pantalla cuántos segundos han pasado. Cuando llega a 5, el ciclo termina y el semáforo cambia a verde. ​

```python
segundos = 0
while segundos < 5:
  print(f"Han pasado {segundos} segundos")
  segundos = segundos + 1
print("El semáforo ha cambiado a verde")
```

>**CICLOS CON BREAK**

```python
print("Programa que determina con tu altura si ingresas a la montaña rusa")

while True:
    estatura = float(input("Ingrese su estatura en metros: "))

    if estatura > 1.34:
        print("Bienvenido a la montaña rusa")
        break
    else:
        print("No puedes ingresar a la montaña rusa")

print("Fin del programa")
```
