**T702 - Ejercicio - Función detectarEtapaVacunacion**

```python
def detectarEtapaVacunacion(edad):
    if edad > 50:
        return 1
    elif edad >= 30:
        return 2
    else:
        return 3
```

**T705 - Ejercicio - Función tipoCohete**


```python
def tipoCohete(cohete):
    carga = cohete["carga"]
    if carga > 50000:
        return "Carga superpesada"
    elif carga > 20000:
        return "Carga pesada"
    elif carga > 2000:
        return "Carga media"
    else:
        return "Carga liviana"
```
